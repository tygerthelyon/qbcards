```python
from qbr.notebook import StockFisherNotebook
```

## 2. Configure

Choose an answer and optional filters. You can also change these in the interface after running the next cell.

```python
ANSWER = "photoelectric effect"
CATEGORY = "Physics"       # e.g. Biology, Architecture; blank for all
DIFFICULTY = ""            # e.g. "6-9"
YEARS = ""                 # e.g. "2018-"
RANKING = "spec"           # "spec", "n", or "pyramid"
MAX_CLUES = 15
```

## 3. Explore

Run this cell, then choose **Find stock clues**. Use the other tabs to search questions or check a draft clue. The mirror selector appears if no database is connected.

```python
dashboard = StockFisherNotebook(
    answer=ANSWER, category=CATEGORY, difficulty=DIFFICULTY,
    year=YEARS, ranking=RANKING, limit=MAX_CLUES,
)
dashboard.display()
```

## 4. Review and export

Inspect the examples and source sets before using a clue. In **Review & export**, select clues and save them to a new CSV file. The optional Anki comparison reads notes through AnkiConnect; it does not edit the collection. The CSV is a research aid, not an Anki deck.
