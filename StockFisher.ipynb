# StockFisher

Find recurring clues in a local QBReader mirror and inspect the questions behind them. For installation and mirror setup, see the [README](README.md#setup).

## 1. Setup

Install the notebook requirements and build or select a mirror. Run the cells from the repository root.

```python
from qbr.notebook import StockFisherNotebook
```

## 2. Configure

Choose an answer and any filters. You can change these in the interface too.

```python
ANSWER = "photoelectric effect"
CATEGORY = "Physics"       # e.g. Biology, Architecture; blank for all
DIFFICULTY = ""            # e.g. "6-9"
YEARS = ""                 # e.g. "2018-"
RANKING = "spec"           # "spec", "n", or "pyramid"
MAX_CLUES = 15
```

## 3. Explore

Run this cell and select **Find stock clues**. Use the tabs to search questions or check a draft clue. If the mirror is not connected, enter its path in the selector.

```python
dashboard = StockFisherNotebook(
    answer=ANSWER, category=CATEGORY, difficulty=DIFFICULTY,
    year=YEARS, ranking=RANKING, limit=MAX_CLUES,
)
dashboard.display()
```

## 4. Review and export

Check the examples and source sets before using a clue. Under **Review & export**, select clues to save as a CSV. The optional Anki comparison reads notes through AnkiConnect without changing them.

The CSV is for research; it is not an Anki deck. After profiling an answer, `dashboard.result` holds the underlying Python data.
