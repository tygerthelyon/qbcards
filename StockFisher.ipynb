Find recurring clues for an answer in QBReader questions. Change the inputs below, then run the cell. The results include question counts and examples from the source sets.

```python
from qbr.notebook import StockFisherNotebook

answer = "photoelectric effect"
category = "Physics"          # leave blank to search every category
difficulty = ""              # e.g. "6-9"
years = ""                   # e.g. "2018-"
rank_by = "spec"             # specificity; use "n" for frequency or "pyramid" for position
clues = 15

research = StockFisherNotebook(
    answer=answer, category=category, difficulty=difficulty,
    year=years, ranking=rank_by, limit=clues,
)
research.run_profile()
research.display()
```

Change the answer or filters in the output and select ?Find clues? to run another search. The other tabs search question text, check a draft clue, and export selected results to CSV.
